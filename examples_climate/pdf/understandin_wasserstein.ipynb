
# Quick 1-Wasserstein demo

We build two very coarse discrete distributions on a 1D grid and use `OptimalTransport.jl` to compute their 1-Wasserstein (Earth mover) distance. The cost matrix is simply the pairwise absolute distance between support points.


```julia
using OptimalTransport

# Toy discrete distributions (support points + weights)
support_a = [0.0, 1.0, 2.0]
weights_a = [0.5, 0.3, 0.2]

support_b = [0.0, 0.5, 3.0]
weights_b = [0.4, 0.4, 0.2]

# Pairwise ground cost: absolute distance on the line
cost_matrix = [abs(x - y) for x in support_a, y in support_b]

wasserstein = emd(weights_a, weights_b, cost_matrix)
println("1-Wasserstein distance: $(wasserstein)")
```
